```python
#W6A16
def in_ma_tran():
    try:
        n, m = map(int, input("Nhập n (hàng) và m (cột): ").split())
        matrix = []
        
        print(f"Nhập {n} dòng, mỗi dòng {m} số:")
        for i in range(n):
            
            row = list(map(int, input().split()))
            matrix.append(row)

        print("Kết quả:")
        for i in range(n):
            for j in range(m):
                
                print(f"{matrix[i][j]:>4}", end="") 
            print() 
    except:
        print("Dữ liệu nhập không hợp lệ.")

```

```python
#W6A17
def in_duong_cheo():
    n = int(input("Nhập kích thước n: "))
    matrix = []
    for i in range(n):
        row = list(map(int, input().split()))
        matrix.append(row)

    print("Đường chéo chính:", end=" ")
    for i in range(n):
        print(matrix[i][i], end=" ")
    
    print("\nĐường chéo phụ:", end=" ")
    for i in range(n):
        print(matrix[i][n - 1 - i], end=" ")
```

```python
#W6A18
def tong_cot_k():
    line1 = list(map(int, input("Nhập n, m, k: ").split()))
    n, m, k = line1[0], line1[1], line1[2]
    
    matrix = []
    for i in range(n):
        matrix.append(list(map(int, input().split())))

    tong = 0
    
    for i in range(n):
        if k < m: 
            tong += matrix[i][k]
            
    print(f"Tổng cột {k}: {tong}")

```

```python
#W6A19
def vi_tri_max(lst):
    if not lst:
        return -1
    
    max_val = lst[0]
    max_index = 0
    
    for i in range(1, len(lst)):
        if lst[i] > max_val:
            max_val = lst[i]
            max_index = i
            
    return max_index

```

```python
#W6A20
def tim_vi_tri_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i 
    return -1 

```

```python
#W6A21
def tao_dict_vi_tri():
    chuoi_nhap = input("Nhập các chuỗi: ").split()
    ket_qua = {}

    for i, tu in enumerate(chuoi_nhap):
        if tu in ket_qua:
            
            ket_qua[tu].append(i)
        else:
            
            ket_qua[tu] = [i]
    
    
    for k in ket_qua:
        ket_qua[k] = tuple(ket_qua[k])
        
    print(ket_qua)

```

```python
#W6A23
def xoa_so_nho_hon_x(lst, x):
    
    list_moi = []
    for num in lst:
        if num >= x:
            list_moi.append(num)
    return list_moi

```

```python
#W6A24
def day_so_0_ve_dau(lst):
    list_so_0 = []
    list_khac = []
    
    for x in lst:
        if x == 0:
            list_so_0.append(x)
        else:
            list_khac.append(x)
            
    return list_so_0 + list_khac

```

```python
#W6A25
def tim_so_thieu(lst):
    print("Các số từ 0-9 bị thiếu là:", end=" ")
    for i in range(10):
        if i not in lst: 
            print(i, end=" ")
    print()

```

```python
#W6A26
def tong_cuc_dai(lst):
    tong = 0
    for i in range(1, len(lst) - 1):
        if lst[i-1] < lst[i] and lst[i] > lst[i+1]:
            tong += lst[i]
    return tong

```

```python
#W6A27
def tim_vi_tri_nho_hon_x(lst, x):
    tim_thay = False
    print(f"Vị trí các số < {x}:", end=" ")
    
    for i in range(len(lst)):
        if lst[i] < x:
            print(i, end=" ")
            tim_thay = True
            
    if not tim_thay:
        print("-1")
    else:
        print() 

```

```python
#W6A28
def kiem_tra_day(lst):
    is_tang = True
    is_giam = True
    
    for i in range(len(lst) - 1):
        if lst[i] >= lst[i+1]:
            is_tang = False
        if lst[i] <= lst[i+1]:
            is_giam = False
            
    if is_tang:
        print("Dãy tăng thực sự")
    elif is_giam:
        print("Dãy giảm thực sự")
    else:
        print("Vô danh")

```
